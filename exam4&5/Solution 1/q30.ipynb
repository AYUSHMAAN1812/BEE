 A 320 KVA, 1/K = 1000/500 V transformer has primary re
sistance and secondary resistance of R1 = 0.048 Ω andR2 =  0.021
 Ω. Calculate the copper loss of the transformer, when
 supplying loadl =  (4/7)th fraction of full rated load

```python
K = 450/1050 ##1/ratio
V2 = 450 ##voltes from ration (denominator)
P = 320*1000 ##VA
R1 = 0.03 ##ohms
R2 = 0.018 ##ohms
loadl = 3/7 ##fraction
Ro1 = R1*K**2
Req = Ro1 + R2
Ifl = P/V2
Cufl = Ifl**2 *(Req)
Cusl = Cufl*(loadl)**2
print("The cu loss at loadl =,",loadl," is ",Cusl)

```
