A W kW, $\phi$ phase, f Hz, P pole induction motor has rotor resistance and reactance at stand still of x Ω and y Ω per phase. The full load torque is obtained at Nr. Find the ratio of maximum torque to full load torque.

```python
phi = 3
f = 50
P = 4
x = 0.11
y = .33
Nr = 1335
S1 = x/y
Ns = 120*f/P
S2 = (Ns-Nr)/Ns
ratio = (S1*(x**2 + (S2*y)**2))/(S2*(x**2 + (S1*y)**2))
print(ratio)
```
