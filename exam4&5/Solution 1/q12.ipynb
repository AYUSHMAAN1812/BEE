A P kVA, V1/V2 V, single phase transformer has the following resistance and reactance parameters on primary and secondary sides: R1 Ω , R2 Ω,X1 Ω , X2 Ω Find the voltage regulation when transformer is supplying full-load cosphi pf lagging.

```python
import numpy as np
V1 = 1600
V2 = 200
k =V2/V1
R1 = 9
R2 = 0.04
X1 = 14
X2 = 0.08
P = 20
R02 = R1*k**2 + R2
X02 = X1*k**2 + X2
Ifl = P*1000/V2
cosphi = 0.77
sinphi = np.sqrt(1-cosphi**2)
Vdrop = Ifl*(R02*cosphi + X02*sinphi)
reg = Vdrop*100/V2
print(reg)
```
