An ideal transformer has a turns ratio of N1:N2. The
following connections are made:
Primary side: Voltage of V1 V is applied
Secondary side: a resistance of R kΩ is connected.
What is the input power taken by the primary side?

```python
N1= 12
N2 = 125
V1 = 210
R = 13
V2 = N2*V1/N1
P = V2**2/(R*1000)
print(P)
```
