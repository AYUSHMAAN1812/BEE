A $\phi$-phase, f Hz, P Pole induction motor has a rotor impedance of (x + jy) ohm at stand still. If the full load torque is obtained at N rpm, Calculate the ratio of starting torque to full load torque

```python
Nr = 627.75
f = 45
P = 8
Ns = 120*f/P
S = (Ns-Nr)/Ns
x = .02
y = .17
ratio = (x**2 + (S*y)**2)/(S*(x**2 + (y)**2))
print(ratio)
```
