. A 6 kVA, 450/120 V transformer has equivalent resis
tance refered to primary, Req(HV) = 2.34 Ω and equiv
alent reactance referred to primay, Xeq(HV) = 3.28 Ω.
 A lagging load is being supplied by it resulting in the
 following meter readings (meters are placed on the HV
 side): I1 = 8.3 A, V1 = 450 V, P1 = 2.5 kW For this
 condition calculate what a voltmeter would read if con
nected across the load terminals. Neglect the no-load
 current of the transformer.

```python
import numpy as np
k = 120/500 ##1/ratio
Req=2.1 ##ohms
Xeq = 4.06 ##ohms
P1 = 3.5*1000 ##W
I1 = 9.2 ##A
V1 = 500 ##V
cosphi = P1/(V1*I1)
sinphi = np.sqrt(1-cosphi**2)

dv = I1*(Req*cosphi+Xeq*sinphi)
E1 = V1 - dv
V2 = E1*k
print(V2 ," and ", E1)
```
