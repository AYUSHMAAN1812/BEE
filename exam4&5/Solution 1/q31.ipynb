A a = 4 kV/400 V, P = 410 kVA transformer has a primary
 winding resistance of R1 = 0.6 Ω and a secondary winding
 resistance of R2 = 0.006 Ω. The iron loss is Wi = 3.75 kW. If the
 power factor of the load is pf = 0.85 lag, determine the effi
ciency of the transformer on full load

```python
a = 4000/400 ##ratio
V2 = 400 ##v ratio denominator
P = 410*1000 ##VA
R1 = 0.6 ##ohms
R2=0.006 ##ohms
Wi = 3.75*1000 ##W
pf = 0.85
I = P/(V2) 
Ro1 = R1/(a**2)
Req = Ro1+R2
Cul = I**2 *(Req)
Eff  =  P*pf/( P*pf + Cul+Wi)
print("The efficiency is ",Eff)
```
