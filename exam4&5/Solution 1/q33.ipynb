Calculate the regulation of the transformer in which
 ohmic loss is 1% of the output and the reactance drop
 is 5% of the voltage when the power factor is 0.8 lag

```python
Ol = 1/100 ##percent
Rd = 5/100 ##percent
cosphi = 0.8 ##pf,lag
sinphi = (1-cosphi**2)**0.5
Reg = Ol*cosphi + Rd*sinphi
print("The regulation is ,",Reg)
```
