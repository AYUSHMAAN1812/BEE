The power input to the rotor of a $\phi$-phase, f Hz, P pole induction motor is Pin kW, the rotor copper losses are Pcr_prephase W per phase. Calculate the speed of rotor in rpm, ignoring the mechanical losses.

```python
phi = 3
f = 40
P = 4
Pin = 12
Pcr_perphase = 240
Pcr = Pcr_perphase*phi
Ns = 120*f/P
S = Pcr/(Pin*1000)
Nr = Ns*(1-S)
print(Nr)
```
