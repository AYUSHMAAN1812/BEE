Asingle-phase transformer has the maximum efficiency
 of 98%. The core losses are Wi = 235 W and the equiva
lent winding resistance as seen from the primary side
 is R01 = 0.6 Ω. The rated current on the primary side is Ifl =  32 A.
 At what percentage of the rated input current does the
 maximum efficiency occur?

```python
Wi = 235 ##w
Ro1 = 0.6 ##ohms
Ifl = 32 ##A
Ic = (Wi/Ro1)**0.5
perc = Ic/Ifl*100
print("The percentage is", perc)
```
