The parameters of a P KVA, V1/V2 V f Hz transformer are: $\\$
Note: All parameters are referred to the primary side. $\\$
Equivalent resistance = $E_{res}$ Ω $\\$
Equivalent reactance = $E_{react}$ Ω $\\$
Core loss resistance = $C_{res}$ Ω $\\$
Magnetization reactance = $M_{react}$ Ω $\\$

As the general practice and from practical convenience point of view, which side of the transformer should the meters be connected for performing the OC test and SC Test, respectively?

OC test --> LV side $\\$
SC test --> HV side

What is the input power taken from the supply during
the OC test?

```python
import numpy as np
P = 4.3
V1 = 480
V2 = 240

Eres = 0.263
Ereact = 0.326
Cres = 708
Mreact = 384
Iw = V2/Cres
Imu = V2/Mreact
Io = np.sqrt(Iw**2+Imu**2)
Poc = Iw**2 * Cres
print(Poc)
```

What is the input power taken from the supply during the SC test?

```python
k = V1/V2 if V1 > V2 else V2/V1
Ros = Eres*k**2
# print(Ros)
Xos = Ereact*k**2
Zos = np.sqrt(Ros**2+Xos**2)
Isc = P*1000/V1
# print(Isc)
Vsc = Isc*Zos
Psc = Isc**2 * Ros
print(Psc)
```
