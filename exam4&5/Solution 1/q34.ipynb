 A 360/170 V transformer has an iron loss of 934 W
 when excited from high voltage side under no load con
dition. If the no load current is 11.28 A, then the no
 load power factor is

```python
Wi = 934 ##W
Inl = 11.28 ##A
V = 360 ##Vnumerator
pf = Wi/(Inl*V)
print("The power fector at no load is ,",pf)
```
