```python
import math

# ---------- given data ----------
# Transformer ratings
V_lv_rated = 220.0    # V, low‐voltage side rated voltage
V_hv_rated = 400.0    # V, high‐voltage side rated voltage

# Open‐circuit (OC) test (measured on LV side)
V_oc = 220.0          # V
I_oc = 0.7            # A
P_oc = 50.0           # W

# Short‐circuit (SC) test (measured on HV side)
V_sc = 16.0           # V
I_sc = 12.0           # A
P_sc = 90.0           # W

# ---------- calculations ----------

# 1) Core‐loss branch on LV side
R0 = V_oc**2 / P_oc
I_c = P_oc / V_oc
I_mu = math.sqrt(I_oc**2 - I_c**2)
X0 = V_oc / I_mu

# 2) Series impedance from SC test (on HV side)
Z_eq_hv = V_sc / I_sc
R_eq_hv = P_sc / I_sc**2
X_eq_hv = math.sqrt(Z_eq_hv**2 - R_eq_hv**2)

# 3) Reflect series resistance to LV side
a = V_lv_rated / V_hv_rated              # turns (voltage) ratio N1/N2 = V1/V2
R_eq_lv = R_eq_hv * a**2

# ---------- results ----------
print(f"27. R01 (series R referred to LV)    = {R_eq_lv:.3f} Ω")
print(f"28. X0  (magnetizing reactance LV)    = {X0:.0f} Ω")
print(f"29. X02 (series X referred to HV)    = {X_eq_hv:.3f} Ω")
print(f"30. R0  (core‐loss R on LV side)     = {R0:.0f} Ω")
```
