```python
# Input parameters (modifiable)
primary_turns = 12
secondary_turns = 135
primary_voltage = 190  # volts
secondary_resistance = 10000  # ohms

# Calculate voltage ratio and secondary voltage
voltage_ratio = secondary_turns / primary_turns
secondary_voltage = primary_voltage * voltage_ratio

# Calculate secondary power (which equals primary power in ideal transformer)
secondary_power = (secondary_voltage ** 2) / secondary_resistance

# Input power (same as secondary power for ideal transformer)
input_power = secondary_power

print(f"Input power: {input_power:.0f} W")
```
