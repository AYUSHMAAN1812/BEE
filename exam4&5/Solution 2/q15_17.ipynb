```python
# Given parameters (modify these values for different scenarios)
KVA_rating = 25  # kVA
N1 = 390         # Primary turns
N2 = 50          # Secondary turns
V1 = 2700        # Primary voltage (V)
freq = 40        # Frequency (Hz)

# Question 16: Secondary induced EMF
E2 = V1 * (N2 / N1)
print(f"16. Secondary EMF: {E2:.2f} V")

# Question 15: Full load secondary current
I2 = (KVA_rating * 1000) / E2
print(f"15. Secondary current: {I2:.2f} A")

# Question 17: Maximum flux in core
phi_max = (V1) / (4.44 * freq * N1) * 1000  # Convert to mWb
print(f"17. Maximum flux: {phi_max:.2f} mWb")
```
