```python
kVA = 20
V1 = 1800
V2 = 290
R1 = 6
R2 = 0.05
X1 = 13
X2 = 0.11
pf = 0.52

# Equivalent impedance calculation
turns_ratio = V1 / V2
Req = R1 + R2 * (turns_ratio ** 2)
Xeq = X1 + X2 * (turns_ratio ** 2)

# Full-load current
I_fl = (kVA * 1000) / V1

# Power factor components
sin_phi = (1 - pf ** 2) ** 0.5

# Voltage regulation
regulation = (I_fl * (Req * pf + Xeq * sin_phi) / V1) * 100
print(f"Answer 18: {regulation:.3f}%")  # Output: 11.633%
```

```python
ohmic_loss_pct = 1
reactance_drop_pct = 3
pf = 0.8

sin_phi = (1 - pf ** 2) ** 0.5
regulation = (ohmic_loss_pct * pf) + (reactance_drop_pct * sin_phi)
print(f"Answer 19: {regulation:.2f}%")  # Output: 2.60%
```

```python
V_primary_line = 210  # Delta connection (phase voltage = line voltage)
V_secondary_line = 14000  # Star connection (phase voltage = line / √3)
power = 40000
pf = 0.82

# Apparent power
S = power / pf

# Secondary line current
I_secondary_line = S / (3 ** 0.5 * V_secondary_line)

# Turns ratio (primary phase voltage / secondary phase voltage)
turns_ratio = V_primary_line / (V_secondary_line / (3 ** 0.5))

# Primary phase current (delta winding)
I_primary_phase = I_secondary_line / turns_ratio
print(f"Answer 20: {I_primary_phase:.1f} A")  # Output: 77.4 A
```
