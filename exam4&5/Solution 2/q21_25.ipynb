```python
import math

# ---------- given data (all referred to primary side) ----------
S_rated   = 4.5e3     # VA, transformer rating
V_pri     = 480.0     # V, primary (high‐voltage) side rated voltage
V_sec     = 210.0     # V, secondary (low‐voltage) side rated voltage
f         = 50.0      # Hz

# Equivalent circuit parameters (all referred to primary side)
R_eq_pri  = 0.26      # Ω, series resistance
X_eq_pri  = 0.329     # Ω, series reactance
R0_pri    = 680.0     # Ω, core‐loss branch resistance
Xm_pri    = 372.0     # Ω, magnetizing reactance

# ---------- test configuration choices ----------
# 21) meter connection sides
oc_test_side = 'LV'   # OC test on the LV side (for convenience)
sc_test_side = 'HV'   # SC test on the HV side

# ---------- calculations ----------

# turns (voltage) ratio for reflecting to LV side
a = V_sec / V_pri     # N_sec / N_pri

# 1) Open‐circuit test (on LV side)
V_oc      = V_sec        # voltage applied during OC test
# reflect core branch impedances to LV side
R0_lv     = R0_pri * a**2
Xm_lv     = Xm_pri * a**2

# OC test meter readings
P_oc      = V_oc**2 / R0_lv                             # real power absorbed
I_c       = V_oc / R0_lv                                # core (loss) current
I_m       = V_oc / Xm_lv                                # magnetizing current
I_oc      = math.sqrt(I_c**2 + I_m**2)                  # total no‐load current

# 2) Short‐circuit test (on HV side)
I_sc      = S_rated / V_pri                             # rated current on HV side
Z_eq_pri  = math.sqrt(R_eq_pri**2 + X_eq_pri**2)        # magnitude of series impedance

# SC test meter readings
P_sc      = I_sc**2 * R_eq_pri                          # real power absorbed
V_sc      = I_sc * Z_eq_pri                             # voltage required to circulate I_sc

# ---------- results ----------
print("21. OC test on:", oc_test_side, "side; SC test on:", sc_test_side, "side")
print(f"22. Input power during OC test  = {P_oc:.1f} W")
print(f"23. Input power during SC test  = {P_sc:.1f} W")
print(f"24. Supply voltage during SC    = {V_sc:.1f} V")
print(f"25. Current drawn during OC test= {I_oc:.3f} A")
```
